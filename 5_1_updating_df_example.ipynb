```python
import pandas as pd
people = {
    "first": ["Corney", "Jane", "Dima"], 
    "second": ["Spiler", "Doe", "Markow"], 
    "email" : ["ssd@", "sdfd@", "gbbhn@"]
}
df = pd.DataFrame(people)
df
```

```python
df.loc[[1, 2], "second"]
```

```python
df.columns
```

```python
df.columns = ['first_name', 'second_name', 'email']
df
```

```python
df.columns = [col.upper() for col in df.columns]
df
```

```python
df.columns = df.columns.str.replace('_', '-')
df
```

```python
df.columns = [col.lower() for col in df.columns]
df
```

```python
df.rename(columns={'first-name': 'first', 'second-name':'last'}, inplace=True)
df
```

# Updating date in the DF

```python
df.loc[2, 'email'] = 'dima@gmail.com'
df
```

```python
df.loc[0] = ['Adriana', 'G', 'adrG@gmalcom']
df
```

```python
df.loc[1, ['last', 'email']] = ['Lans', 'jlans@gmail.com'] #or df.loc[filt_fuctions, 'name_of_col'] = ...
df
```

```python
df.at[1, 'last'] = 'Green' #probably faster then loc
df
```

# Updating date columns

```python
df['email'] = df['email'].str.lower()
df
```

## apply(func) -> used for each row 

#### with Series

```python
df['email'].apply(len)
```

```python
df = df['email'].apply(lambda email : email.upper())
```

```python
df['email'] = df['email'].str.lower()
df
```

#### with DataFrames

```python
df.apply(len) #used for each column by default
```

```python
df.apply(len, axis='columns') #used for each row
```

```python
df.apply(pd.Series.min)
```
